```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    def __str__(self):
        return f'{self.name}, {self.age}'
    
p1 = Person("Caesar", 76)

print(p1)
```

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    def BlowNewJerseyUp(self):
        print(f"A florida man known as {self.name}, {self.age}, was last seen on West 62nd Street after reports came in of him attempting to \"Blow New Jersey Up\"")

p1 = Person("Harry", 26)
p1.BlowNewJerseyUp()
```

```python
class IceWizard:
    def __init__(self, hasCrown, hasIntelligence, hatesGunter):
        self.hasCrown = hasCrown
        self.hasIntelligence = hasIntelligence
        self.hatesGunter = hatesGunter
    
    def slapsGunter(self):
        print("Bad Gunter *slaps Gunter in the face* Bad Gunter")
```

```python
class Gunter(IceWizard):
    pass

myGunter = Gunter(False, False, True)
myGunter.slapsGunter()
```
